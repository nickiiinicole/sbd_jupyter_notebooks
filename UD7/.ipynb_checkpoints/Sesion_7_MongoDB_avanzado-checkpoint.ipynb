**Replicaset (replicación)**

**Concepto**: conjunto de nodos que mantienen copias idénticas de los mismos datos.

**Objetivo**: alta disponibilidad y tolerancia a fallos.

1. Un nodo es **primario**, permite **escritura**.

2. Los demás son **secundarios** y **replican datos del primario**.

3. Si el primario falla, **uno de los secundarios se convierte en primario** automáticamente.

Esto permite protege tus datos de pérdida y permite lecturas distribuidas.

Ahora vamos a parar mongodb1 y comprobar que otro coge el papel de primario. Podemos hacerlo con Docker

A continuación nos conectamos, por ejemplo, mongodb2

*docker exec -it mongodb2 mongo*

Y comprobamos que cual es el primario y podemos hacer una inserción"

*rs.status()*


```python
from pymongo import MongoClient
from pymongo.errors import OperationFailure

# --- Conexión al primario (mongodb2) ---
cliente_primario = MongoClient('mongodb2', 27017,
                               replicaset='rs0',
                               readPreference='primary')

db_primario = cliente_primario['bbdd']
coleccion_primario = db_primario['compras']

# Intentamos insertar un documento
doc_primario = {"producto": "TestPrimario", "cantidad": 5}

resultado_primario = coleccion_primario.insert_one(doc_primario)
print(f"Inserción en primario OK, id: {resultado_primario.inserted_id}")


# --- Conexión al secundario (mongodb3) ---
cliente_secundario = MongoClient('mongodb3', 27017,
                                 replicaset='rs0',
                                 readPreference='secondary')

db_secundario = cliente_secundario['bbdd']
coleccion_secundario = db_secundario['compras']

# Intentamos insertar un documento en secundario
doc_secundario = {"producto": "TestSecundario", "cantidad": 10}

try:
    resultado_secundario = coleccion_secundario.insert_one(doc_secundario)
    print(f"Inserción en secundario OK, id: {resultado_secundario.inserted_id}")
except OperationFailure as e:
    print(f"No se pudo insertar en secundario: {e}")
```

**Muy importante**, lo que está pasando no es que haya insertado en el secundario sino que dirigió la inserción al primario, pero si hago una conexión direrecat sin replicaset así:

```python
from pymongo import MongoClient
from pymongo.errors import OperationFailure

# Conexión directa al secundario dentro del replicaset e intentar escribir
cliente_sec = MongoClient(
    'mongodb3', 
    27017, 
    replicaset='rs0',     # indicamos replicaset
    directConnection=True  # fuerza conexión directa a este nodo
)
db_sec = cliente_sec['bbdd']
coleccion_sec = db_sec['compras']

try:
    coleccion_sec.insert_one({"test": 123})
except OperationFailure as e:
    print("Fallo al intentar escribir directamente en secundario:", e)
```

# Sharding

El **sharding** es un mecanismo de **particionado horizontal de datos**

En lugar de tener todos los datos en un único servidor, **MongoDB los divide y reparte entre varios servidores, llamados shards**.

Cada **shard almacena solo una parte del total de los datos.**

Se usa cuando una base de datos empieza a tener problemas de:

1. Volumen de datos (demasiados documentos)
2. Rendimiento (muchas lecturas/escrituras concurrentes)
3. Límites físicos de un servidor (CPU, RAM, disco)

Con sharding conseguimos:

**Escalar añadiendo máquinas**, no haciendo una más grande

**Repartir carga** de lectura y escritura

**Mantener tiempos de respuesta** estables aunque crezcan los datos

Entoces ¿que ventajas tengo frente a replica-set?

 
1. **ReplicaSet**: Alta disponibilidad	**Sharding**: Escalabilidad
2. **Replicaset**: Copias de los mismos datos	**Sharding**: Datos repartidos
3. **ReplicaSet**: Un primario escribe	**Sharding**: Varios shards escriben
4. **ReplicaSet**: Protege ante caídas	**Sharding**: Permite crecer

**IMPORTANTE** NO SON EXCLUYENTES. EN PRODUCCIÓN SE CONFIGURAN AMBOS

Componentes del sharding en MongoDB

Un **clúster sharded** tiene **3 piezas fundamentales**:

1. **Shards**

Donde están los datos

Cada shard guarda solo una parte

Normalmente: replica sets

2. **Config Servers**

Guardan la metainformación

Qué shard tiene qué datos

Dónde empieza y acaba cada rango

No guardan datos de usuario

3. **mongos (router)**

Es el punto de entrada

La aplicación se conecta siempre a mongos

Decide a qué shard enviar cada operación

**Para la app, parece una sola base de datos**


La **shard key** es el campo (o campos) de la bbdda que MongoDB usa para repartir los datos. Una mala elección afecta al rendimiento del cluster.

Tipos de sharding
1. **Sharding por rango (range-based)**

Los datos se dividen por intervalos. Ej: fechas, precios, ids
- Bueno para consultas por rango
- Riesgo de “hot shard”

2. **Sharding por hash**

MongoDB hace un hash del valor. Distribución uniforme

- Muy equilibrado
- Peor para consultas por rango

**¿Cuándo NO usar sharding?**

1. Bases de datos pequeñas

2. Pocos usuarios

3. Aplicaciones sencillas

4. Cuando un replica set es suficiente

**¿por qué? Sharding añade complejidad**:  más nodos, más configuración, más puntos de fallo.

**IDEAS CLAVE**

1. **Replica Set** = seguridad
2. **Sharding** = crecimiento

Y siempre:

**Primero replica set, luego sharding**

## Conexión a mongos

```python
from pymongo import MongoClient

#Utiliza la IP de tu anfitrión
cliente = MongoClient('mongos', 27017)

# comprobamos que es mongos
if cliente.admin.command("hello").get("msg") == "isdbgrid":
    print("OK: conectado a mongos")


#Creamos la instancia para interactuar con la colección de clientes y pedidos
bbdd = cliente.tienda
coleccion_clientes = bbdd.clientes
coleccion_pedidos = bbdd.pedidos

#shardeamos si es necesario si no pass

try:
    cliente.admin.command({
        "shardCollection": "tienda.clientes",
        "key": {"dni": 1}
    })
    cliente.admin.command({
        "shardCollection": "tienda.pedidos",
        "key": {"dni": 1}
    })
except:
    pass

#ANTES APLIQUEMOS ENABLE SHARDING A LA BBDD Y A LA COLECCIÓN
```

## CONCEPTOS CLAVE

1. La bbdd la veo en mongos, vive *logicamente* en mongos, **pero mongos luego reparte los datos**

2. mongos es un **router**, no una base de datos, no almacena nada, consulta **config servers** y decide **a que shard va cada chunk y a que nodo manda** cada operación

3. Ves la BBDD en mongos pero **no la ves necesariamente** en mongodb1, mongodb2, mongodb3. Eso no significa que estos no puedan tener bbdd adicionales.

4. Cuando insertas, **mongos** mira el shard key, consulta **configsvr** y decide *este dni va al shard rs0*, envía la escritura al primary de rs0 y, por último, entra en acción **replicaset** que replica el documento en sus secundarios.

5. Regla: **replicaset = copia**, **sharding = divide**, **mongos = director de orquesta**

6. La bbdd se ve en mongos, los datos se guardan en los shard o resto de nodos

   

* mongos es el director de orquesta.

* Los músicos son los shards (ReplicaSets)

* La partitura son los config servers

* El público son las aplicaciones (Jupyter, Compass, backend…)

* El público solo ve al director, nunca va músico por músico.
