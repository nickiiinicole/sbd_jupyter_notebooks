**Conexion Servido MongoDB**

```python
from pymongo import MongoClient
from pymongo.errors import ConnectionFailure

#conexion a MongoDB desde Docker y el puerto mapeado

try: 
    client = MongoClient("mongodb://172.18.0.6:27017")
    print(client.list_database_names())
    print("conexion realizada")
except: 
    print("Error conexion ")

```

**Creación de una base de datos**

    

```python
db = client['database']
# crear una coleccion de la base de datos 
coleccion = db['usuarios']

# insertar un documento, pq si no lo insertas no crea la bd
usuario = {
    'nombre':'María Sol',
    'email':'maria@gmail.com',
    'edad': 42
}

result = coleccion.insert_one(usuario)
print(f"Documento insertado con ID:{result.inserted_id}")

print(client.list_database_names())
print(client.db.list_collection_names())
```
