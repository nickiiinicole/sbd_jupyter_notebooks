El **preprocesado** es la fase en la que convertimos los datos “crudos” en algo limpio, consistente y listo para analizar. Esto es crucial porque un dataset sucio puede dar resultados engañosos. Básicamente:

1. **Valores nulos o faltantes**

 * Columnas o filas con datos vacíos (NaN en Pandas).

 * Decidir si rellenar (fillna) o eliminar (dropna).

Ejemplo: edades que faltan en un dataset de usuarios.

2. **Duplicados**

* Filas repetidas que pueden distorsionar estadísticas.

* Detectar con duplicated() y eliminar con drop_duplicates().

3. **Tipos de datos incorrectos**

* Por ejemplo, fechas como strings, números como texto.

* Convertir con astype() o pd.to_datetime().

4. **Valores inconsistentes o anómalos**

* Valores fuera de rango, errores de tipeo, diferentes formatos de lo mismo.

* Ejemplo: “M” / “Masculino” / “Hombre” → unificar.

4. **Normalización o estandarización de columnas**

* Nombres de columnas consistentes (lower_case, sin espacios).

* Formato homogéneo en strings: mayúsculas, minúsculas, remover espacios.

6. **Preparación de variables derivadas**

* Crear nuevas columnas que faciliten el análisis:

* Edad a partir de fecha de nacimiento.

* Categorías agrupadas.

Mientras el **procesado de datos** es la fase en la que, a partir de datos limpios, se realizan transformaciones, cálculos y agregaciones para extraer información útil, detectar patrones y preparar resultados para análisis o visualización. 


### **1. Preprocesado**

Antes tengamos en cuentas dos conceptos básicos de **Pandas**:

Un **dataset** es un conjunto de datos organizados en filas y columnas, donde:

Cada **fila** representa un registro u observación (por ejemplo, un usuario, una venta, un coche).

Cada **columna** representa una característica o variable (por ejemplo, edad, nombre, precio).

En Pandas, los datasets se representan con DataFrames.



##### **EXPLORACIÓN INICIAL**

```python
import pandas as pd
import numpy as np

# Crear un DataFrame de ejemplo
data = {
    'nombre': ['Ana', 'Juan', 'Luis', 'Jorge', 'Pilar', 'María', 'Ana'],
    'edad': [23, 29, 44, np.nan, 34, 71, 23],
    'genero': ['F', 'M', 'M', None, 'F', 'F', 'F'],
    'ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla', None, 'Bilbao', 'Madrid'],
    'ingresos': [2500, 3200, 4100, 2800, 3000, None, 2500]
}

df = pd.DataFrame(data)

# Mostrar el DataFrame
display(df)
```

#### **Selección y Filtrado**

```python
# Selección de columna (Series)
edades = df['edad']
print(edades)
# Selección de varias columnas (DataFrame)
nombres_ciudades = df[['nombre', 'ciudad']]
print(nombres_ciudades)

# Filtrar filas: personas mayores de 30 años
mayores_30 = df[df['edad'] > 30]
print(mayores_30)

# Selección por posición
fila_0 = df.iloc[0]

print(fila_0)

# Selección por etiquetas
fila_jorge = df.loc[df['nombre'] == 'Jorge']

print(fila_jorge)
```

#### **Valores Nulos**

Detectar y tratar datos faltantes

```python
df.isnull().sum()      # cantidad de nulos por columna
df.isna().sum()        # equivalente a isnull()
```

```python
df.dropna(inplace=True)   # Eliminar filas con nulos
display(df)
```

```python
## la volvemos a crear

# Crear un DataFrame de ejemplo
data = {
    'nombre': ['Ana', 'Juan', 'Luis', 'Jorge', 'Pilar', 'María', 'Ana'],
    'edad': [23, 29, 44, np.nan, 34, 71, 23],
    'genero': ['F', 'M', 'M', None, 'F', 'F', 'F'],
    'ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla', None, 'Bilbao', 'Madrid'],
    'ingresos': [2500, 3200, 4100, 2800, 3000, None, 2500]
}

df = pd.DataFrame(data)

# Mostrar el DataFrame
display(df)
```

Rellenamos nulos

```python
df['edad'].fillna(df['edad'].median(), inplace=True)
df['genero'].fillna('Desconocido', inplace=True)
df['ciudad'].fillna('Desconocida', inplace=True)
df['ingresos'].fillna(df['ingresos'].mean(), inplace=True)

display(df)
```

#### **Duplicados**

```python
df.duplicated()      # Detectar duplicados

df.duplicated().sum()  # Contar duplicados
```

Eliminamos duplicados 

```python
df.drop_duplicates(inplace=True) 

display(df)
```

#### **Modificación de tipos**

```python

df['ingresos'] = df['ingresos'].astype(int)  # Convertir ingresos a entero

# Si tuviéramos fecha, se usaría:
# df['fecha'] = pd.to_datetime(df['fecha'])

display(df)
```

```python
df['nombre'] = df['nombre'].str.lower().str.strip() # Pasar nombres a minúsculas y quitar espacios


df['genero'] = df['genero'].replace({'F': 'Femenino', 'M': 'Masculino'}) # Unificar categorías

display(df)
```

#### **Variables derivadas**

```python

df['mayor_de_edad'] = df['edad'] >= 18  # Crear columna mayor de edad


df['grupo_edad'] = pd.cut(df['edad'], bins=[0, 18, 35, 60, 100],
                           labels=['Joven', 'Adulto', 'Maduro', 'Mayor'])  # Crear grupo de edad

display(df)
```

#### **Otras operaciones de preprocesado** (hay muchos más)

Renombrar índices 

```python

# Mostrar índice original
print("Índice original:")
print(df.index)

# Reiniciar índice si hubiesemos eliminado una fila volvería a reinciar
df.index = range(len(df))

# Mostrar DataFrame con nuevo índice
print("\nDataFrame con índice reiniciado:")
display(df)
```

Eliminar columnas irrelevantes

```python
# Añadir columna irrelevante
df['columna_irrelevante'] = ['irrelevante'] * len(df) #para saber el número de filas

# Mostrar el DataFrame con la nueva columna
print("DataFrame con columna irrelevante:")
display(df)

# Eliminar la columna irrelevante
df.drop(columns=['columna_irrelevante'], inplace=True)

# Mostrar el DataFrame tras eliminar la columna
print("DataFrame tras eliminar la columna irrelevante:")
display(df)
```

Transformación numérica

```python
# Normalizar la columna 'edad' cuanto se desvía cada valor
df['edad_norm'] = (df['edad'] - df['edad'].mean()) / df['edad'].std()

# Mostrar el DataFrame con la nueva columna
display(df)
```

Detectar valores extremos

```python
# Valor mínimo y máximo de la columna 'edad'
edad_min = df['edad'].min()
edad_max = df['edad'].max()

print("Edad mínima:", edad_min)
print("Edad máxima:", edad_max)

# Filtrar filas que son exactamente los valores extremos
extremos = df[(df['edad'] == edad_min) | (df['edad'] == edad_max)]
display(extremos)
```

Transformación de categorías

```python
# Antes de la transformación
display(df['genero'])

# Reemplazamos abreviaturas por valores completos
df['genero'] = df['genero'].replace({'F': 'Femenino', 'M': 'Masculino', None: 'Desconocido'})

# Después de la transformación
display(df['genero'])
```

Combinar columnas (muy útil a veces)

```python
# Por ejemplo, sumar ingresos y edad solo a modo de ejemplo
df['edad_por_ingresos'] = df['edad'] * df['ingresos']
display(df[['edad', 'ingresos', 'edad_por_ingresos']])
```

Un caso más util sería si queremos unir nombre y apellidos

### **2. Procesado**

El procesado es la fase que sigue al preprocesado y consiste en trabajar con datos ya limpios y consistentes para extraer información útil, calcular métricas y preparar resultados para análisis o visualización.

1 **Objetivo**

* Transformar los datos limpios en información significativa.
* Detectar patrones, relaciones o tendencias.
* Preparar datos para visualización, informes o modelos de análisis.

2. Diferencia con el preprocesado

 * Preprocesado: limpiar, normalizar y preparar los datos (eliminar nulos, duplicados, corregir tipos, etc.).
 * Procesado: transformar esos datos para extraer conocimiento mediante cálculos, agregaciones, agrupaciones y métricas.

 3. Qué se hace en esta fase:

 * Filtrar y seleccionar datos relevantes.
 * Agrupar y resumir información (groupby, tablas dinámicas).
 * Crear métricas o columnas derivadas (por ejemplo, ingresos totales o ratios).
 * **Preparar datos para visualizar o exportar.**

```python
import pandas as pd
import numpy as np

# DataFrame de ejemplo completo
data = {
    'nombre': ['Ana', 'Juan', 'Luis', 'Jorge', 'Pilar', 'María', 'Ana', 'Luis'],
    'edad': [23, 29, 44, 38, 34, 71, 23, 44],
    'genero': ['Femenino', 'Masculino', 'Masculino', 'Masculino', 'Femenino', 'Femenino', 'Femenino', 'Masculino'],
    'ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla', 'Madrid', 'Bilbao', 'Madrid', 'Valencia'],
    'ingresos': [2500, 3200, 4100, 2800, 3000, 5000, 2500, 4100],
    'bonus': [200, 150, 300, 180, 220, 400, 200, 300]
}

df = pd.DataFrame(data)

# Mostrar el DataFrame
display(df)
```

**Selección y filtrado**

```python
# Seleccionar la columna 'edad' (devuelve una Series)
edades = df['edad']
display(edades)

# Seleccionar varias columnas (devuelve un DataFrame)
datos_basicos = df[['nombre', 'edad', 'genero']]
display(datos_basicos)

# Selección por posición (iloc)
primera_fila = df.iloc[0]
display(primera_fila)

# Selección por etiquetas (loc)
fila_jorge = df.loc[df['nombre'] == 'Jorge']
display(fila_jorge)
```

```python
# Personas mayores de 30 años
mayores_30 = df[df['edad'] > 30]
display(mayores_30)

# Usando query() (más legible)
mayores_30_q = df.query('edad > 30')
display(mayores_30_q)
```

**Agrupaciones y agregaciones**  Importante

```python
# Media de ingresos por género
media_ingresos_genero = df.groupby('genero')['ingresos'].mean()
display(media_ingresos_genero)
```

```python
# Edad y bonus: media, máximo y mínimo por ciudad
resumen_ciudad = df.groupby('ciudad').agg({
    'edad': ['mean', 'max', 'min'],
    'bonus': ['mean', 'sum']
})
display(resumen_ciudad)
```

```python
# Cuántas personas hay en cada ciudad
conteo_ciudad = df.groupby('ciudad')['nombre'].count()
display(conteo_ciudad)
```

**Ordenar Datos**

```python
#ORDENAR POR UNA COLUMNA
# Ordenar el DataFrame por edad de menor a mayor
df_orden_edad = df.sort_values(by='edad', ascending=True)
display(df_orden_edad)

# Ordenar por ingresos de mayor a menor
df_orden_ingresos = df.sort_values(by='ingresos', ascending=False)
display(df_orden_ingresos)
```

```python
#ORDENAR POR VARIAS
# Ordenar primero por ciudad ALFABETICO y luego por ingresos descendente
df_orden_multiple = df.sort_values(by=['ciudad', 'ingresos'], ascending=[True, False])
display(df_orden_multiple)
```

**Resumen de Datos**

```python
# Estadísticas básicas de columnas numéricas
resumen_numerico = df.describe()
display(resumen_numerico)
```

```python
#Conteo categorías 

# Contar cuántas personas hay por ciudad
conteo_ciudad = df['ciudad'].value_counts()
display(conteo_ciudad)

# Contar cuántas personas hay por género
conteo_genero = df['genero'].value_counts()
display(conteo_genero)
```

**Creación de Nuevas Métridas**

```python
#crear una columna con ratio en este caso ingresos por edad

df['ingreso_por_edad'] = df['ingresos'] / df['edad']
display(df[['nombre', 'edad', 'ingresos', 'ingreso_por_edad']])
```

```python
#crear una columna categórica MUY IMPORTANTE

df['categoria_ingresos'] = pd.cut(df['ingresos'],
                                   bins=[0, 3000, 4000, 6000],
                                   labels=['Bajo', 'Medio', 'Alto'])
display(df[['nombre', 'ingresos', 'categoria_ingresos']])

#pd.cut() permite agrupar valores numéricos en rangos y etiquetarlos.
```

**Pivot y tablas dinámicas**

```python
tabla = pd.pivot_table(df, 
                       values='ingresos',  # columna que queremos resumir
                       index='genero',     # filas de la tabla
                       columns='categoria_ingresos',  # columnas de la tabla
                       aggfunc='mean')     # función de agregación
display(tabla)
```

```python
#Con varias métricas 
tabla2 = pd.pivot_table(df,
                        values=['ingresos', 'bonus'],
                        index='ciudad',
                        aggfunc={'ingresos':'mean', 'bonus':'sum'})
display(tabla2)
```

**Merge y Join**

```python
# DataFrame adicional 
df_bonus = pd.DataFrame({
    'nombre': ['Ana', 'Juan', 'Luis'],
    'bonus_extra': [50, 100, 75]
})

# Combinar con el DataFrame principal
df_merged = pd.merge(df, df_bonus, on='nombre', how='left')
display(df_merged)

#how='inner' → solo filas que coinciden en ambos DataFrames.

#how='right' → mantiene todas las filas del segundo DataFrame.

#how='outer' → combina todas las filas de ambos DataFrames, llenando con NaN donde no hay coincidencia.
```

**Fitrado condiciones múltiles**

```python
#pseronas con más de 30 años e ingresos mayores de 3000
filtro = df[(df['edad'] > 30) & (df['ingresos'] > 3000)]
display(filtro)
```

```python
# menores de 25 o con ingresos mayores de 4000

filtro_or = df[(df['edad'] < 25) | (df['ingresos'] > 4000)]
display(filtro_or)
```

```python
# los que no son de Madrid
filtro_not = df[~(df['ciudad'] == 'Madrid')]
display(filtro_not)
```

**Top N valores**

```python

#ver los N mayores y menores resgistros por si son "absurdos"

top_ingresos = df.nlargest(3, 'ingresos')
display(top_ingresos)
```

```python
# los N menores

menores_edad = df.nsmallest(3, 'edad')
display(menores_edad)
```

**Binning o agrupación de rangos**

```python

# Crear rangos de edad
df['grupo_edad'] = pd.cut(df['edad'], 
                          bins=[0, 30, 50, 100], 
                          labels=['Joven', 'Adulto', 'Mayor'])
display(df[['nombre', 'edad', 'grupo_edad']])
```

```python
#cuantas hay por grupos 
conteo_grupo = df['grupo_edad'].value_counts()
display(conteo_grupo)
```

**Codificación Binaria**

Básica para categorizar variables de dos categorías

```python
df['genero_binario'] = df['genero'].map({'Masculino':1, 'Femenino': 0})
display(df)
```

**One-Hot Encoding** (variables dummy)

Para variables con más de dos categorías (ciudad, categoría_ingresos, grupo_edad…).

```python
dummies_ciudad = pd.get_dummies(df['ciudad'], prefix='city')

# Añadirlas al mismo DataFrame
df = pd.concat([df, dummies_ciudad], axis=1)

display(df)
```

**Normalización Numérica (Min-Max)**

Convierte los valores a un rango 0–1. Util cuando tienen escalas distintas y muy usado en redes neuronales.

La **normalización** consiste en transformar valores numéricos a una escala común para que todas las columnas tengan rangos comparables. Por ejemplo, imagina estas dos variables:

Edad: entre 18 y 90

Ingresos: entre 1.000 y 50.000

Si las metes en un modelo tal cual los ingresos dominarán cualquier cálculo, porque tienen números mucho más grandes.
Si hacemos normalización numérica se mejora la precisión de los algoritmos, especialmente el entrenamiento de redes neuronales, y se pueden comparar variables en antes **estaban en escalas diferentes**. Hay varias técnicas pero solo vermos la más usada.


```python
#lo haremos con edad  e ingresos
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[['edad_norm', 'ingresos_norm']] = scaler.fit_transform(df[['edad', 'ingresos']])

display(df[['nombre', 'edad', 'edad_norm', 'ingresos', 'ingresos_norm']])
```

**Estandarización (Z-score)**

Convierte los valores para que tengan media = 0 y desviación típica = 1. Persigue el mismo fin que la normalización numérica.

```python
from sklearn.preprocessing import StandardScaler
std = StandardScaler()
df[['edad_std', 'ingresos_std']] = std.fit_transform(df[['edad', 'ingresos']])

display(df[['nombre', 'edad', 'edad_std', 'ingresos', 'ingresos_std']])
```
