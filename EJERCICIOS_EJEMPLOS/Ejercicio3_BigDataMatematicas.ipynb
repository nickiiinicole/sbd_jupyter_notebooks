#### 3. Calcula la distribución de Poisson y Normal del apartado anterior.

Calcula el número de días que superan los 1000 euros de ventas y la probabilidad de haya más de 5 días con más de 1000 euros por día.

Nota.- Para calcular el número de días con más 1000 euros ventas  ====> dias_1000 = (ventas > 1000).sum()

Muestra las gráficas correspondientes y explícalas

#### Distribución Poison

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

ventas = np.array([1057, 662, 1999, 1690, 1367, 931, 1628, 1254, 1608, 1942, 746, 120, 940, 266, 1397, 487, 700, 415, 113, 341, 876, 1469, 664, 997, 1463, 191, 1490, 1055, 1578, 1551])
dias_1000 = (ventas > 1000).sum()
lam = dias_1000  # media de ventas por día
x = np.arange(0,30)
y = poisson.pmf(x, lam)
plt.bar(x,y)
plt.title('Probabilidad de que las ventas por día superen los 1000 euros en un mes')
plt.xlabel('Días')
plt.ylabel('Probabilidad')
plt.show()
```

#### Distribución Normal

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
from scipy.stats import norm
mu, sigma = 1033.23, 546.72  # media de altura y desviación estándar
# mu = media de la distribución (promedio de la altura, en este caso).

# sigma = desviación estándar (qué tan dispersos están los datos respecto a la media).
x_norm = np.linspace(140, 200, 30)     #array de 100 puntos entres 140 y 200
y_norm = norm.pdf(x_norm, mu, sigma)   #calcula la función de densidad de probabilidad (PDF) de la distribución normal.
plt.plot(x_norm, y_norm)
plt.title('Distribución Normal (altura)')
plt.xlabel('Ventas')
plt.ylabel('Densidad de probabilidad')
plt.show()
```
