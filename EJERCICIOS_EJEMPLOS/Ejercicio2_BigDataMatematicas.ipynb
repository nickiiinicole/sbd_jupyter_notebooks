#### 2. Una tienda online registra las ventas diarias de un producto durante 30 días (1 mes) expuestos en el vector siguiente. Calcula media, mediana, moda y varianza.

```python
import numpy as np
from scipy import stats

ventas = np.array([1057, 662, 1999, 1690, 1367, 931, 1628, 1254, 1608, 1942, 746, 120, 940, 266, 1397, 487, 700, 415, 113, 341, 876, 1469, 664, 997, 1463, 191, 1490, 1055, 1578, 1551])

# Media
media = round(np.mean(ventas), 2)

# Mediana
mediana = round(np.median(ventas), 2)

# Moda 
moda = round(stats.mode(ventas, keepdims=True).mode[0], 2)

# Varianza
varianza = round(np.var(ventas), 2)

# Desviación
desviacion = round(np.std(ventas), 2)

print('Media:', media)
print('Mediana:', mediana)
print('Moda:', moda)
print('Varianza:', varianza)
print ('Desviación estandar:', desviacion)
```
